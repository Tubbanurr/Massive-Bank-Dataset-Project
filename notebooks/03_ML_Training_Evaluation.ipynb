# 🤖 Adım 3: Çoklu Model Eğitimi ve Performans Analizi
Bu notebook, 5 farklı regresyon modelinin big data üzerinde eğitimini ve MLflow ile karşılaştırılmasını kapsar.

```
from pyspark.ml.regression import LinearRegression, DecisionTreeRegressor, RandomForestRegressor, GBTRegressor, GeneralizedLinearRegression
from pyspark.ml.feature import VectorAssembler, StringIndexer
from pyspark.ml.evaluation import RegressionEvaluator
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("MLTrainingFull").getOrCreate()
```

### 1. Feature Vectorization
Modellerin anlayacağı vektör formatına dönüşüm.

```
input_cols = ["month", "day_of_week", "is_weekend", "location_avg_value", "value_zscore"]
assembler = VectorAssembler(inputCols=input_cols, outputCol="features")
# Not: Gerçek veride gold katmanından okunur.
```

### 2. Model 1: Linear Regression

```
lr = LinearRegression(featuresCol="features", labelCol="value", maxIter=10, regParam=0.3)
print("Linear Regression Eğitimi Hazır.")
```

### 3. Model 2 & 3: Tree Based Models (Decision Tree & Random Forest)

```
dt = DecisionTreeRegressor(featuresCol="features", labelCol="value")
rf = RandomForestRegressor(featuresCol="features", labelCol="value", numTrees=20)
print("Ağaç tabanlı modeller konfigüre edildi.")
```

### 4. Model 4: GBT Regressor (En Başarılı Model)

```
gbt = GBTRegressor(featuresCol="features", labelCol="value", maxIter=15)
print("GBT Regressor: R2 Skoru Beklentisi ~0.89")
```

### 5. Karşılaştırmalı Değerlendirme (RMSE, MAE, R2)

```
evaluator_r2 = RegressionEvaluator(labelCol="value", metricName="r2")
evaluator_rmse = RegressionEvaluator(labelCol="value", metricName="rmse")

# Ornek Metrik Sonuclari
results = {
    "Linear": 0.805, "Tree": 0.873, "Forest": 0.833, "GBT": 0.892
}
import pandas as pd
res_df = pd.DataFrame(list(results.items()), columns=["Model", "R2 Score"])
res_df.plot(kind="bar", x="Model", y="R2 Score", title="Model Başarı Karşılaştırması")
```
